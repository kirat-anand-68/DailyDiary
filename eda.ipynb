```python
import pandas as pd
```

```python
df=pd.read_csv("AdventureWorks Sales Data 2022.csv")
df.head()
```

```python
df['StockYear']=df['StockDate'].str.split('-').str[0]
df['StockMonth']=df['StockDate'].str.split('-').str[1]
df['StockDate']=df['StockDate'].str.split('-').str[2]
```

```python
df['OrderYear']=df['OrderDate'].str.split('-').str[0]
df['OrderMonth']=df['OrderDate'].str.split('-').str[1]
df['OrderDate']=df['OrderDate'].str.split('-').str[2]
```

```python
df.info()
```

```python
df.head()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Top Products
top_products = df.groupby('ProductKey')['OrderQuantity'].sum().nlargest(10)
plt.figure(figsize=(10,4))
top_products.plot(kind='bar', color='teal')
plt.title("🏆 Top 10 Products by Order Quantity")
plt.xlabel("ProductKey")
plt.ylabel("Total Quantity")
plt.show()
```

```python
top_customers = df.groupby('CustomerKey')['OrderQuantity'].sum().nlargest(10)
plt.figure(figsize=(10,4))
top_customers.plot(kind='bar', color='orange')
plt.title("👤 Top 10 Customers by Order Quantity")
plt.xlabel("CustomerKey")
plt.ylabel("Total Quantity")
plt.show()
```

```python
# 6️⃣ Relationship between Dates
# ======================
plt.figure(figsize=(8,5))
sns.scatterplot(x='OrderDate', y='StockDate', data=df, color='red')
plt.title("Order Date vs Stock Date")
plt.xlabel("Order Date")
plt.ylabel("Stock Date")
plt.show()
```

```python
sns.pairplot(data=df)
```

```python
sns.barplot(y='OrderQuantity',x='OrderMonth',data=df)
plt.xlabel("Order Month")
plt.ylabel("Order Quantity")
plt.title("Order Quantity per Month")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['CustomerKey'], bins=20, kde=True, color='orange')
plt.title("Number of customers")
plt.show()
```

```python
sales1=df.groupby('')['Total Revenue'].sum()
```

```python
plt.figure(figsize=(10,4))
sns.heatmap(df.isnull(), cbar=False, cmap='viridis')
plt.title("Missing Values Heatmap")
plt.show()
```
